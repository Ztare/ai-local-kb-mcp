# Knowledge base — MCP tools (full)

You work with a knowledge base (KB) of articles.

**Terms**
- **domain** — top-level topic (e.g. `coding`); contains only topics.
- **topic** — `coding/python`; **subtopic** — `coding/python/packaging` (max 3 levels).
- **article** — text about one question inside a topic: `coding/python/asyncio`.
- **section** — titled part of an article: `coding/python/asyncio#event-loop`. The first section of an article (`#head`) is its title and short description.
- **tag** — a label of a section that links it to other places outside the tree (`event loop`).

**Paths**
- `""` — the root (list of domains); `domain/topic[/subtopic]` — a topic; `.../article` — an article; `article#id` — a section.
- **Copy paths from previous answers. Never invent them.** The only path you compose yourself is the one of a new object (parent path + new name).
- Names and ids: only `a-z`, `0-9`, `-`.

**Answers and errors**
- Every write answer says `created` or `updated`. `created` when you meant to update means a typo in the path.
- An error always has a hint (the next call to make) and, when possible, a list of valid options. Pick from the options; do not retry the same call.

**How to work**
1. Find: `search` / `search_regex` by text; tags: `search_section_tags*` → `sections_by_tag`. Walk the tree with `get` / `index`.
2. Read: `get(article)` → outline; `get(section)` → text; `full=true` for the whole article.
3. Write: `save_topic` / `save_article` / `save_section`; tags: `add_section_tags` / `drop_section_tags`; remove: `delete`.


## index

Flat list of all articles (path — title) in the whole KB or inside a topic, at any depth. Parameters:
- `path` (optional) — a domain or topic to limit the list.


## get

Read by path. Parameters:
- `path` (required) — copy it from previous answers.
- `full` (default `false`) — for an article: return all sections with text instead of the outline.
- `with_tags` (default `false`) — add section tags.

What you get:
- `""` → list of domains;
- topic → its head + subtopics and articles (path — title);
- article → outline: head + section paths and titles (or everything with `full=true`);
- section → title and text.


## search

Find sections whose title or text contains **any** of the substrings. Parameters:
- `substrings` (required) — list of exact substrings, case does not matter. Add variants: `["задач", "task"]`.
- `path` (optional) — limit the search to a domain, topic or article.

Use short parts of words to catch word forms. Returns section paths with titles. Nothing found is not an error — try shorter or other substrings.


## search_regex

Like `search`, but with a regular expression (Python syntax). Parameters:
- `pattern` (required);
- `path` (optional).

A bad pattern returns `invalid_pattern` with the parser message.


## section_tags

Unique tags of all sections inside a domain, topic or article. Parameters:
- `path` (required) — tags of the whole KB are not available; start with a domain.


## search_section_tags

Find tags that contain **any** of the substrings. Parameters:
- `substrings` (required) — case does not matter;
- `path` (optional).

Use it before tagging, to reuse existing tags, and before `sections_by_tag`, to get the exact tag.


## search_section_tags_regex

Like `search_section_tags`, but with a regular expression. Parameters: `pattern` (required), `path` (optional).


## sections_by_tag

All sections with exactly this tag. Parameters:
- `tag` (required) — copy it from `section_tags` / `search_section_tags`;
- `path` (optional);
- `with_tags` (default `false`).


## save_topic

Create a topic or replace its head. Parameters:
- `path` (required) — parent path + new name: `coding/python`;
- `title` (required) — human name;
- `content` (required) — short description: what this topic is for.

The parent must exist; at most 3 levels.


## save_article

Create an article or replace its head. Parameters:
- `path` (required) — topic path + new name: `coding/python/asyncio`;
- `title` (required);
- `content` (required) — short description of the article.

Articles live in topics and subtopics, not directly in a domain. The topic must exist.


## save_section

Create or replace a section. Parameters:
- `path` (required) — `article#id`. New id → the section is added to the **end** of the article; existing id → it is replaced.
- `title` (required) — unique within the article.
- `content` (required) — the text without the title line.

Always send the **full** text of the section: whatever you send replaces the old text. To extend a section, read it with `get` first.


## add_section_tags

Add tags to a section. Parameters:
- `path` (required) — `article#id`;
- `tags` (required) — list or comma-separated string.

Reuse existing tags (find them with `search_section_tags`) instead of inventing synonyms. Adding an existing tag is not an error.


## drop_section_tags

Remove tags from a section. Parameters: `path` (required), `tags` (required). Removing a missing tag is not an error.


## delete

Delete a section, an article or an **empty** topic. Parameters:
- `path` (required).

A head cannot be deleted on its own — delete the whole article or topic.
