# База знаний — инструменты MCP (full)

Ты работаешь с базой знаний (БЗ) из статей.

**Термины**
- **область** — тема верхнего уровня (например `coding`); внутри только темы.
- **тема** — `coding/python`; **подтема** — `coding/python/packaging` (не больше 3 уровней).
- **статья** — текст по одному вопросу внутри темы: `coding/python/asyncio`.
- **секция** — часть статьи с заголовком: `coding/python/asyncio#event-loop`. Первая секция статьи (`#head`) — её название и краткое описание.
- **тег** — метка секции, связывающая её с другими местами вне дерева (`event loop`).

**Пути**
- `""` — корень (список областей); `область/тема[/подтема]` — тема; `.../статья` — статья; `статья#id` — секция.
- **Копируй пути из прошлых ответов. Не выдумывай их.** Сам составляешь только путь нового объекта (путь родителя + новое имя).
- Имена и id: только `a-z`, `0-9`, `-`.

**Ответы и ошибки**
- Ответ на запись всегда говорит `created` (создано) или `updated` (обновлено). `created` там, где ты хотел обновить, — значит, опечатка в пути.
- В ошибке всегда есть подсказка (какой вызов сделать дальше) и, где можно, список допустимых вариантов. Выбирай из вариантов, не повторяй тот же вызов.

**Как работать**
1. Найти: `search` / `search_regex` по тексту; по тегам: `search_section_tags*` → `sections_by_tag`. По дереву — `get` / `index`.
2. Прочитать: `get(статья)` → оглавление; `get(секция)` → текст; `full=true` — статья целиком.
3. Записать: `save_topic` / `save_article` / `save_section`; теги: `add_section_tags` / `drop_section_tags`; удалить: `delete`.


## index

Плоский список всех статей (путь — название) во всей базе или внутри темы, на всю глубину. Параметры:
- `path` (необязателен) — область или тема, чтобы ограничить список.


## get

Чтение по пути. Параметры:
- `path` (обязателен) — копируй из прошлых ответов.
- `full` (по умолчанию `false`) — для статьи: все секции с текстом вместо оглавления.
- `with_tags` (по умолчанию `false`) — добавить теги секций.

Что вернётся:
- `""` → список областей;
- тема → её шапка + подтемы и статьи (путь — название);
- статья → оглавление: шапка + пути и заголовки секций (или всё при `full=true`);
- секция → заголовок и текст.


## search

Найти секции, в заголовке или тексте которых есть **любая** из подстрок. Параметры:
- `substrings` (обязателен) — список точных подстрок, регистр не важен. Добавляй варианты: `["задач", "task"]`.
- `path` (необязателен) — искать только внутри области, темы или статьи.

Бери части слов, чтобы поймать словоформы. Возвращает пути секций с заголовками. Ничего не найдено — не ошибка: попробуй подстроки короче или другие.


## search_regex

Как `search`, но по регулярному выражению (синтаксис Python). Параметры:
- `pattern` (обязателен);
- `path` (необязателен).

Некорректное выражение — ошибка `invalid_pattern` с текстом ошибки разбора.


## section_tags

Уникальные теги всех секций внутри области, темы или статьи. Параметры:
- `path` (обязателен) — теги по всей базе не отдаются; начни с области.


## search_section_tags

Найти теги, в которых есть **любая** из подстрок. Параметры:
- `substrings` (обязателен) — регистр не важен;
- `path` (необязателен).

Вызывай перед тем, как ставить теги, — чтобы взять существующие, и перед `sections_by_tag` — чтобы получить точный тег.


## search_section_tags_regex

Как `search_section_tags`, но по регулярному выражению. Параметры: `pattern` (обязателен), `path` (необязателен).


## sections_by_tag

Все секции с точно таким тегом. Параметры:
- `tag` (обязателен) — копируй из `section_tags` / `search_section_tags`;
- `path` (необязателен);
- `with_tags` (по умолчанию `false`).


## save_topic

Создать тему или заменить её шапку. Параметры:
- `path` (обязателен) — путь родителя + новое имя: `coding/python`;
- `title` (обязателен) — человеческое название;
- `content` (обязателен) — краткое описание: для чего эта тема.

Родитель должен существовать; не больше 3 уровней.


## save_article

Создать статью или заменить её шапку. Параметры:
- `path` (обязателен) — путь темы + новое имя: `coding/python/asyncio`;
- `title` (обязателен);
- `content` (обязателен) — краткое описание статьи.

Статьи лежат в темах и подтемах, не прямо в области. Тема должна существовать.


## save_section

Создать или заменить секцию. Параметры:
- `path` (обязателен) — `статья#id`. Новый id → секция добавляется в **конец** статьи; существующий → заменяется.
- `title` (обязателен) — уникален в статье.
- `content` (обязателен) — текст без строки заголовка.

Всегда отправляй **полный** текст секции: отправленное заменяет старое. Чтобы дополнить секцию, сначала прочитай её через `get`.


## add_section_tags

Добавить теги секции. Параметры:
- `path` (обязателен) — `статья#id`;
- `tags` (обязателен) — список или строка через запятую.

Бери существующие теги (найди через `search_section_tags`), не выдумывай синонимы. Повторное добавление — не ошибка.


## drop_section_tags

Убрать теги секции. Параметры: `path` (обязателен), `tags` (обязателен). Убрать отсутствующий тег — не ошибка.


## delete

Удалить секцию, статью или **пустую** тему. Параметры:
- `path` (обязателен).

Шапку отдельно удалить нельзя — только всю статью или тему.
