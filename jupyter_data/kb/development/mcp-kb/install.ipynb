# Чистая установка

Как поставить плагин базы знаний на новую машину или телефон без чужих данных: что нужно заранее, какие файлы перенести, а какие нет, и порядок запуска. Рабочую базу при этом не чистят — переносится только выборка файлов.

## Что нужно заранее

- **Windows:** Python 3.11 (запускается как `py -3.11`; другой — `set PY=путь\к\python.exe`).
- **Android:** Termux; один раз `pkg install python rust clang make libzmq`. По желанию termux-api — чтобы процессы не умирали при выключенном экране.
- Интернет на первый запуск: скрипт ставит `jupyterlab` и `dulwich` через pip.

## Какие файлы перенести

Проще всего взять репозиторий проекта: в нём ровно список ниже (его задаёт корневой `.gitignore`). Пути — от корня проекта; структуру каталогов сохранить.

| Перенести | Зачем |
|---|---|
| `kb_mcp/` | сам плагин |
| `start_jupyter.bat`, `start_jupyter.sh`, `README.md`, `LICENSE` | запуск, описание, лицензия |
| `jupyter_data/config/jupyter_server_config.json` | включённый плагин, ip, порт |
| `jupyter_data/config/lab/user-settings/@jupyterlab/notebook-extension/tracker.jupyterlab-settings` | запуск и остановка ядра без вопросов (`development/mcp-kb/save-prompts`) |
| `jupyter_data/config/lab/user-settings/@jupyterlab/apputils-extension/notification.jupyterlab-settings` | не ходить за новостями Jupyter |
| `jupyter_data/mcp_docs/` | описания инструментов и правила индексатора (без них — только встроенные английские для админки) |
| `jupyter_data/kb/development/topic.ipynb` и `jupyter_data/kb/development/mcp-kb/` | документация самого плагина и навык `kb-index` — они живут только в базе |
| `tests/` | по желанию, для разработки |

**Не переносить:**
- `jupyter_bin/` — ставится сам для своей ОС (`win` / `android`);
- `jupyter_data/config/token.txt` — личный токен, на новом месте плагин создаст свой;
- остальные настройки Lab в `user-settings/` — личные привычки; `jupyter_data/config/lab/workspaces/`, `jupyter_data/config/migrated`, `jupyter_data/data/`, `jupyter_data/runtime/` — состояние JupyterLab (открытые вкладки и т.п.), создаётся само;
- `jupyter_data/kb/.git/` и `.gitignore` — история автокоммитов старой базы; плагин создаст новый репозиторий сам;
- другие области базы (`jupyter_data/kb/games/` и т.п.) — личные знания;
- `.ipynb_checkpoints/`, `.idea/`, `.pytest_cache/`, `__pycache__/`, `.claude/` — служебные.

## Установка по шагам

1. Поставить Python (см. «Что нужно заранее»).
2. Скопировать файлы из списка в любую папку.
3. Запустить `start_jupyter.bat` (Windows) или `sh start_jupyter.sh` (Termux). Первый запуск дольше — установка пакетов в `jupyter_bin/`.
4. Плагин сам создаст токен (`jupyter_data/config/token.txt`), git-репозиторий базы и напечатает в консоль адреса и готовые команды `claude mcp add …`.
5. Подключить ИИ: скопировать команду из консоли и выполнить в каталоге нужного проекта; для индексатора — ещё `kb-admin` и «установи навык из `development/mcp-kb/skills/kb-index`» (подробно — `development/mcp-kb/connect`).

**Проверка:** в браузере открывается адрес `[kb] web:`; ИИ на `get("")` видит одну область `development`. Нет папки `kb/` вообще — не ошибка: плагин создаст пустую базу (но тогда не будет документации и навыка).

## Перенести свои знания

Если нужна не чистая, а своя база на новом месте — скопировать нужные области целиком (`jupyter_data/kb/<область>/`) до или после первого запуска. При старте сервер сверит хеши: неизменённые секции останутся проверенными, переиндексация не нужна. Весь каталог `jupyter_data/kb/` с `.git/` — если нужна и история правок.

Рабочую базу перед переносом чистить не нужно: для чистой установки просто не копируются лишние области.
