# MCP база знаний

Документация проекта Ztare AI Local KB MCP — плагина Jupyter Server, который даёт единую базу знаний: ИИ работает с ней через MCP (/mcp/kb, /mcp/kb-admin), человек читает и правит её в JupyterLab. Цели, хранение, API, формат ответов, устройство плагина, запуск и настройки, чистая установка, подключение к ИИ, работа с базой на телефоне, открытие статей без вопросов о сохранении и навыки для агентов. Начинать со статьи `development/mcp-kb/overview`; поставить на новое место — `development/mcp-kb/install`; подключить новый проект — `development/mcp-kb/connect`.
