# MCP база знаний

Плагин Jupyter Server, который даёт единую базу знаний: ИИ работает с ней через MCP (/mcp/kb, /mcp/kb-admin), человек читает и правит её в JupyterLab. Цели, хранение, API, формат ответов, устройство плагина, запуск, чистая установка, подключение к ИИ и навыки для агентов. Начинать со статьи `development/mcp-kb/overview`; поставить на новое место — `development/mcp-kb/install`; подключить новый проект — `development/mcp-kb/connect`.
